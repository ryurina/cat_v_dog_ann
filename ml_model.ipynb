### 1 Load Image

```python
from PIL import Image
import numpy as np, pandas as pd, os

def flatten_pixels(path, size=(32,32)):
    img = Image.open(path).convert('L').resize(size)
    return np.array(img).flatten()

rows = []
for label, folder in [('cat','./training_set/cats'), ('dog','./training_set/dogs')]:
    for f in os.listdir(folder):
        px = flatten_pixels(os.path.join(folder, f))
        rows.append(list(px) + [label])

cols = [f'pixel_{i}' for i in range(32*32)] + ['label']
df = pd.DataFrame(rows, columns=cols)
```
